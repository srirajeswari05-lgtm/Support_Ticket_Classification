```python
# Import libraries

import pandas as pd
import re
```

```python
# Load preprocessed dataset

df = pd.read_csv("../data/preprocessed_support_tickets.csv")

print("Dataset loaded successfully!")
```

```python
# Check ticket categories

print(df["Topic_group"].value_counts())
```

```python
# Create priority classification function

def assign_priority(text):
    text = text.lower()

    high_keywords = [
        "urgent",
        "critical",
        "down",
        "outage",
        "cannot access",
        "not working",
        "security",
        "data loss",
        "system failure"
    ]

    medium_keywords = [
        "error",
        "issue",
        "problem",
        "failed",
        "slow",
        "unable"
    ]

    if any(keyword in text for keyword in high_keywords):
        return "High"

    elif any(keyword in text for keyword in medium_keywords):
        return "Medium"

    else:
        return "Low"
```

```python
# Assign priority to each ticket

df["Priority"] = df["cleaned_text"].apply(assign_priority)

print("Priority assigned successfully!")
```

```python
# Check priority distribution

df["Priority"].value_counts()
```

```python
# Display sample tickets with assigned priority

df[
    ["Document", "Topic_group", "Priority"]
].head(20)
```

```python
# Check priority distribution by ticket category

pd.crosstab(
    df["Topic_group"],
    df["Priority"]
)
```

```python
# Check percentage of each priority

priority_percentage = (
    df["Priority"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

priority_percentage
```

```python
# Save tickets with predicted priority

df.to_csv(
    "../data/final_support_tickets.csv",
    index=False
)

print("Final support ticket dataset saved successfully!")
```
