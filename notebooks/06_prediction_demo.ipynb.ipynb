```python
# Import libraries

import re
import joblib

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
# Load trained models

category_model = joblib.load(
    "../models/linear_svm_ticket_classifier.pkl"
)

tfidf = joblib.load(
    "../models/tfidf_vectorizer.pkl"
)

print("Models loaded successfully!")
```

```python
# Create stopwords list

stop_words = set(stopwords.words("english"))
```

```python
# Create text preprocessing function

def preprocess_ticket(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    tokens = word_tokenize(text)
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    return " ".join(tokens)
```

```python
# Create priority classification function

def assign_priority(text):
    text = text.lower()

    high_keywords = [
        "urgent",
        "critical",
        "down",
        "outage",
        "cannot access",
        "not working",
        "security",
        "data loss",
        "system failure"
    ]

    medium_keywords = [
        "error",
        "issue",
        "problem",
        "failed",
        "slow",
        "unable"
    ]

    if any(keyword in text for keyword in high_keywords):
        return "High"

    elif any(keyword in text for keyword in medium_keywords):
        return "Medium"

    else:
        return "Low"
```

```python
# Enter a new support ticket

new_ticket = input("Enter your support ticket: ")

print("\nTicket received successfully!")
```

```python
# Preprocess the new ticket

cleaned_ticket = preprocess_ticket(new_ticket)

print("Preprocessed ticket:")
print(cleaned_ticket)
```

```python
# Convert ticket into TF-IDF features

ticket_features = tfidf.transform(
    [cleaned_ticket]
)

print("TF-IDF features created successfully!")
```

```python
# Predict ticket category

predicted_category = category_model.predict(
    ticket_features
)[0]

print("Predicted Category:", predicted_category)
```

```python
# Predict ticket priority

predicted_priority = assign_priority(
    cleaned_ticket
)

print("Predicted Priority:", predicted_priority)
```

```python
# Display final prediction

print("\n===== SUPPORT TICKET RESULT =====")
print("Ticket:", new_ticket)
print("Category:", predicted_category)
print("Priority:", predicted_priority)
print("=================================")
```
