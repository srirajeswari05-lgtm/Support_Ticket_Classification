```python
# Import libraries

import pandas as pd
import numpy as np
```

```python
# Load the new support ticket dataset

df = pd.read_csv("../data/all_tickets_processed_improved_v3.csv")

print("Dataset loaded successfully!")
```

```python
# Check dataset size

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Display column names

print(df.columns.tolist())
```

```python
# Display first 5 rows

df.head()
```

```python
# Check dataset information

df.info()
```

```python
# Check missing values

df.isnull().sum()
```

```python
# Check duplicate rows

print("Duplicate rows:", df.duplicated().sum())
```

```python
# Remove duplicate rows

df = df.drop_duplicates()

print("Duplicate rows removed!")
print("Rows remaining:", len(df))
```

```python
# Remove rows with missing values

df = df.dropna()

print("Missing values removed!")
print("Rows remaining:", len(df))
```

```python
# Check missing values again

print(df.isnull().sum())
```

```python
# Check duplicate rows again

print("Duplicate rows:", df.duplicated().sum())
```

```python
# Check unique ticket categories

print("Number of categories:", df["Topic_group"].nunique())
print("\nCategories:")
print(df["Topic_group"].unique())
```

```python
# Check category distribution

df["Topic_group"].value_counts()
```

```python
# Check text length

df["text_length"] = df["Document"].astype(str).str.len()

df["text_length"].describe()
```

```python
# Remove empty ticket text

df = df[df["Document"].astype(str).str.strip() != ""]

print("Empty ticket texts removed!")
print("Rows remaining:", len(df))
```

```python
# Save the cleaned dataset

df.to_csv(
    "../data/cleaned_support_tickets.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
