```python
# Import libraries

import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
# Download NLTK resources

nltk.download("punkt")
nltk.download("stopwords")
nltk.download("punkt_tab")
```

```python
# Load cleaned dataset

df = pd.read_csv("../data/cleaned_support_tickets.csv")

print("Dataset loaded successfully!")
```

```python
# Check text and category columns

print(df[["Document", "Topic_group"]].head())
```

```python
# Create stopwords list

stop_words = set(stopwords.words("english"))

print("Number of stopwords:", len(stop_words))
```

```python
# Create text cleaning function

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words]
    
    return " ".join(tokens)
```

```python
# Apply text preprocessing

df["cleaned_text"] = df["Document"].astype(str).apply(clean_text)

print("Text preprocessing completed!")
```

```python
# Compare original and cleaned text

df[["Document", "cleaned_text"]].head(10)
```

```python
# Check cleaned text length

df["cleaned_text_length"] = df["cleaned_text"].str.len()

df["cleaned_text_length"].describe()
```

```python
# Check empty cleaned texts

print(
    "Empty cleaned texts:",
    (df["cleaned_text"].str.strip() == "").sum()
)
```

```python
# Remove empty cleaned texts

df = df[df["cleaned_text"].str.strip() != ""]

print("Rows remaining:", len(df))
```

```python
# Save preprocessed dataset

df.to_csv(
    "../data/preprocessed_support_tickets.csv",
    index=False
)

print("Preprocessed dataset saved successfully!")
```
